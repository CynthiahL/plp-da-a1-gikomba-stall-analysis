```python
daily_sales = [2300, 1950, 2800, 2100, 3600, 5200, 4100,

              2500, 2200, 3100, 2400, 3900, 5600, 4700] # 14 days, KES

days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"] * 2



# qty = kg in stock, reorder = reorder level (kg), cost/price = KES per kg

stock = {

    "Tomatoes": {"qty": 18, "reorder": 25, "cost": 60, "price": 90},

    "Onions":  {"qty": 40, "reorder": 20, "cost": 70, "price": 100},

    "Sukuma":  {"qty": 9, "reorder": 15, "cost": 30, "price": 50},

    "Potatoes": {"qty": 55, "reorder": 30, "cost": 50, "price": 75},

    "Avocado": {"qty": 12, "reorder": 10, "cost": 80, "price": 140},

}


total_sales = sum(daily_sales)
avg_sales = total_sales / len(daily_sales)
highest_sales = max(daily_sales)
lowest_sales = min(daily_sales)

best_idx = daily_sales.index(highest_sales)
worst_idx = daily_sales.index(lowest_sales)


best_day_name = days[best_idx]
best_week_num = 1 if best_idx < 7 else 2

worst_day_name = days[worst_idx]
worst_week_num = 1 if worst_idx < 7 else 2

print(
    f"Total Sales: KES {total_sales:,.2f}\n"
    f"Average Sales: KES {avg_sales:,.2f}\n"
    f"Best Day: Week {best_week_num} {best_day_name} (KES {highest_sales:,.2f})\n"
    f"Worst Day: Week {worst_week_num} {worst_day_name} (KES {lowest_sales:,.2f})"
)

```

```python
def week_totals(sales):
  #Return total sales for week 1 and week 2.
   return (sum(sales[:7]), sum(sales[7:]))

w1_total, w2_total = week_totals(daily_sales)
growth_pct = ((w2_total - w1_total) / w1_total) * 100

print(
    f"Week Growth Percentage: {growth_pct:.2f}"
)
```

```python
def classify_day(amount, average):
  #Classify daily transaction amount
    if amount > average * 1.2:
        return "Strong"
    elif amount < average * 0.8:
        return "Weak"
    else:
        return "Normal"

sales_categories = {"Strong": 0, "Normal": 0, "Weak": 0}
for amount in daily_sales:
    category = classify_day(amount, avg_sales)
    sales_categories[category] += 1

print (sales_categories)
```

```python
weekday_sales = []
weekend_sales = []

for idx in range(len(daily_sales)):
    if days[idx] in ["Sat", "Sun"]:
        weekend_sales.append(daily_sales[idx])
    else:
        weekday_sales.append(daily_sales[idx])

avg_weekday = sum(weekday_sales) / len(weekday_sales)
avg_weekend = sum(weekend_sales) / len(weekend_sales)
weekend_premium_pct = ((avg_weekend - avg_weekday) / avg_weekday) * 100

print(
      f" Average week day sale is KES{avg_weekday:.2f}\n",
      f"Average week end sale is KES{avg_weekend:.2f}\n",
      f"The difference in weekdays vs weekend percentages is {weekend_premium_pct:.2f}\n"
    )
```

```python
def process_inventory_metrics(stock_dict):
  #Returns a list of products whose quantity is at or below the reorder level and calculates the gross margin percentage
  needs_restock = []
  calculated_margins = {}

  for item, metrics in stock_dict.items():
      if metrics["qty"] <= metrics["reorder"]:
          needs_restock.append(item)

      margin = ((metrics["price"] - metrics["cost"]) / metrics["price"]) * 100
      calculated_margins[item] = margin

  return needs_restock, calculated_margins

reorder_list, margin_profiles = process_inventory_metrics(stock)
```

```python
#Wanjiku's Stall Business Perfomance Report
print("==================================================")
print("           WANJIKU'S STALL PERFORMANCE REPORT      ")
print("==================================================")
print(f"Total 14-Day Sales       : KES {total_sales:,.2f}")
print(f"Daily Sales Average      : KES {avg_sales:,.2f}")
print(f"Best Performing Day     : Week {best_week_num} {best_day_name} (KES {highest_sales:,.2f})")
print(f"Worst Performing Day    : Week {worst_week_num} {worst_day_name} (KES {lowest_sales:,.2f})")
print("--------------------------------------------------")
print("WEEK-ON-WEEK DEVELOPMENT")
print(f"Week 1 Total Revenue    : KES {w1_total:,.2f}")
print(f"Week 2 Total Revenue    : KES {w2_total:,.2f}")
print(f"Net Growth Rate          : {growth_pct:.1f}%")
print("--------------------------------------------------")
print("SALES DISTRIBUTION DYNAMICS")
print(f"Strong Revenue Days      : {sales_categories['Strong']} days")
print(f"Normal Revenue Days      : {sales_categories['Normal']} days")
print(f"Weak Revenue Days        : {sales_categories['Weak']} days")
print(f"Weekday Average Sales    : KES {avg_weekday:,.2f}")
print(f"Weekend Average Sales    : KES {avg_weekend:,.2f}")
print(f"Weekend Performance Shift: +{weekend_premium_pct:.1f}% higher than weekdays")
print("--------------------------------------------------")
print("INVENTORY & PROFIT MARGIN DIAGNOSTICS")
print(f"CRITICAL REORDER ALERTS  : {', '.join(reorder_list) if reorder_list else 'None'}")
print("Product Profit Margins   :")
for item, margin_val in margin_profiles.items():
    print(f"  - {item:<12} : {margin_val:.1f}% Margin")
print("==================================================")
```
